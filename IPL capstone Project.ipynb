### Loading the Libraries and Dataset

```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
import warnings 
warnings.filterwarnings("ignore")
df = pd.read_csv("IPL.csv")
df.head()
```

```python
match_wins = df["match_winner"].value_counts()
match_wins
```

## Who wins the most Matches?

```python
sns.barplot(y = match_wins.index , x = match_wins.values , palette = "rainbow")
plt.title("Most Matchs Winner Teams")
```

## Top Bowlers

```python
df["highest_wickets"] = df["best_bowling_figure"].apply(lambda x:x.split("--")[0])
df["highest_wickets"] = df["highest_wickets"].astype(int)
top_bowlers = df.groupby("best_bowling")["highest_wickets"].sum().sort_values(ascending = False)
top_bowlers
```

```python
sns.barplot(x = top_bowlers.values , y = top_bowlers.index , palette = "rainbow")
```
